```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.patch.set_facecolor('#fafafa')

# ── Left: polynomial with all real roots ──────────────────────────────────────
ax1 = axes[0]
roots_real = np.array([-4, -2, -0.5, 1, 3])
P_real = np.poly1d(np.poly(roots_real))
x = np.linspace(-5.5, 4.5, 500)
y = P_real(x)

ax1.axhline(0, color='gray', linewidth=0.8, linestyle='--')
ax1.axvline(0, color='gray', linewidth=0.8, linestyle='--')
ax1.plot(x, y, color='#2c7bb6', linewidth=2)
ax1.scatter(roots_real, np.zeros_like(roots_real),
            color='#d7191c', s=80, zorder=5, label='Real roots')
ax1.set_ylim(-60, 80)
ax1.set_title(r'All roots real — $a_k^2 \geq 4\,a_{k-1}a_{k+1}$',
              fontsize=12, pad=12)
ax1.set_xlabel('x', fontsize=11)
ax1.set_ylabel('P(x)', fontsize=11)
ax1.legend(fontsize=10)
ax1.set_facecolor('#f7f7f7')
ax1.spines[['top','right']].set_visible(False)

# ── Right: complex plane — real vs complex roots ──────────────────────────────
ax2 = axes[1]

# Polynomial with 2 complex roots: (x+1)(x+2)(x²+x+3)
coeffs_mixed = np.polymul([1,1,3], np.poly([-1,-2]))
P_mixed = np.poly1d(coeffs_mixed)
roots_mixed = np.roots(coeffs_mixed)

real_roots = roots_mixed[np.abs(roots_mixed.imag) < 1e-8].real
complex_roots = roots_mixed[np.abs(roots_mixed.imag) >= 1e-8]

ax2.axhline(0, color='gray', linewidth=0.8, linestyle='--')
ax2.axvline(0, color='gray', linewidth=0.8, linestyle='--')

ax2.scatter(real_roots, np.zeros_like(real_roots),
            color='#d7191c', s=80, zorder=5, label='Real roots')
ax2.scatter(complex_roots.real, complex_roots.imag,
            color='#1a9641', marker='^', s=90, zorder=5, label='Complex roots')

# Dashed lines from complex roots to real axis
for r in complex_roots:
    ax2.plot([r.real, r.real], [0, r.imag],
             color='#1a9641', linewidth=0.9, linestyle=':')

ax2.set_xlim(-3.5, 1.5)
ax2.set_ylim(-2.5, 2.5)
ax2.set_title(r'Mixed roots — Newton satisfied, Kurtz violated',
              fontsize=12, pad=12)
ax2.set_xlabel('Re', fontsize=11)
ax2.set_ylabel('Im', fontsize=11)
ax2.legend(fontsize=10)
ax2.set_facecolor('#f7f7f7')
ax2.spines[['top','right']].set_visible(False)

plt.suptitle('Real Roots of Real Polynomials', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
#plt.savefig('assets/img/polynomials_roots.png', dpi=150, bbox_inches='tight')
#plt.savefig('../img/polynomials_roots.png', dpi=150, bbox_inches='tight')
plt.show()
```
